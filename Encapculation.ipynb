```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks   # private variable (encapsulation)

    def calculate_average(self,avg):
        self.avg = avg
        total = sum(self.__marks)
        avg = total / len(self.__marks)
        return avg

    def show_details(self):
        return f"Student: {self.name}, Average Marks: {self.calculate_average()}"

```

```python
s1 = Student("Rahul", [80, 90, 85])
print(s1.show_details())

```
