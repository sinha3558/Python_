Student Marks (Multiple Inheritance)

✔️ Yahan MRO (Method Resolution Order) ka use nahi hua
✔️ Kyunki humne explicitly parent constructors call kiye
✔️ Isliye order exact wahi hoga jo hum likhenge

```python
class internalmarks:

  def __init__(self,imarks):
    self.imarks = 30
    print("Internal Marks assigned")

class ExternalMarks:

  def __init__(self,emarks):
    self.emarks = 40
    print("External Marks Assigned")

class Student(internalmarks,ExternalMarks):

  def __init__(self):
    print("Student Record Created")

  def total_marks(self):
    return self.imarks + self.emarks
```

```python
s1 = Student()
print(s1.total_marks())
```

```python
class internalmarks:

  def __init__(self,imarks):
    self.imarks = 30
    print("Internal Marks assigned")

class ExternalMarks:

  def __init__(self,emarks):
    self.emarks = 40
    print("External Marks Assigned")

class Student(internalmarks,ExternalMarks):

  def __init__(self):
    internalmarks.__init__(self,imarks)
    ExternalMarks.__init__(self,emarks)
    print("Student Record Created")

  def total_marks(self):
    return self.imarks + self.emarks
```

```python
class InternalMarks:
    def __init__(self, imarks):
        self.imarks = imarks
        print("Internal Marks assigned")

class ExternalMarks:
    def __init__(self, emarks):
        self.emarks = emarks
        print("External Marks Assigned")

class Student(InternalMarks, ExternalMarks):
    def __init__(self, imarks, emarks):
        InternalMarks.__init__(self, imarks)
        ExternalMarks.__init__(self, emarks)
        print("Student Record Created")

    def total_marks(self):
        return self.imarks + self.emarks
```

```python
s1 = Student(50,89)
print(s1.total_marks())
```

```python
class internalmarks:

  def __init__(self,imarks,*args, **kwargs):
    self.imarks = imarks
    print("internalmraks created")
    super().__init__(*args, **kwargs)

  def get_marks(self):
    return self.imarks

class ExternalMarks:

  def __init__(self,emarks,*args, **kwargs):
    self.emarks = emarks
    print("ExternalMarks Created")
    super().__init__(*args, **kwargs)

  def get_marks(self):
    return self.emarks

class Students(internalmarks,ExternalMarks):
  def __init__(self,imarks,emarks):
    print("Students Record created")
    super().__init__(imarks,emarks)

  def get_marks(self):
    return self.imarks + self.emarks
```

```python
s = Students(78,90)
print(s.get_marks())
```
