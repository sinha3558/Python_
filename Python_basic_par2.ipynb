```python
def calculator(a, b):
    add = a + b
    sub = a - b
    mul = a * b
    div = a / b
    avg = (a + b) / 2
    return add, sub, mul, div, avg


print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
print("5. Average")

select = int(input("Enter a number 1,2,3,4,59: "))

a = int(input("enter the first number : "))
b = int(input("enter the second number : "))

add, sub, mul, div, avg = calculator(a, b)

if select == 1:
    print(f"Sum of two numbers is : {add}")

elif select == 2:
    print(f"Subtraction of two numbers is : {sub}")

elif select == 3:
    print(f"Multiplication of two numbers is : {mul}")

elif select == 4:
    print(f"Division of two numbers is : {div}")

elif select == 5:
    print(f"Average of two numbers is : {avg}")

else:
    print("Invalid input")
```

```python
def sub1(*arg):
  result = arg[0]
  for i in arg[1:]:
    result -= i
  return result

n = int(input(f"enter a number: "))
m1 = []

for i in range(n):
  m2 = int(input(f"enter the {i} number : "))
  m1.append(m2)
print(m1)

result = sub1(*m1)
print(result)
```
